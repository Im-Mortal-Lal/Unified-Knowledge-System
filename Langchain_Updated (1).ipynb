```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_ollama import OllamaLLM
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load PDF file
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("policy.pdf")
docs = loader.load()
```

```python
# Split
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks = splitter.split_documents(docs)

#| Parameter          | Meaning                                  |
#| ------------------ | ---------------------------------------- |
#| `chunk_size=500`   | each chunk contains ~500 characters      |
#| `chunk_overlap=50` | 50 characters repeated in the next chunk |

#Chunk 1 → characters 1–500
#Chunk 2 → characters 450–950
#Chunk 3 → characters 900–1400

```

```python
# Embeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
# Vector DB
vectorstore = FAISS.from_documents(chunks, embeddings)
```

```python
# Retriever
retriever = vectorstore.as_retriever()
```

```python
# LLM
llm = OllamaLLM(model="qwen2.5:3b")
```

```python
query = str(input())
```

```python
docs = retriever.invoke(query)
```

```python
context = "\n".join([d.page_content for d in docs])
```

```python
prompt = f"""understand the Query of the user and answer from the policy document accordingly
Answer the question using the context.

Context:
{context}

Question:
{query}

Answer:
"""
```

```python
response = llm.invoke(prompt)
```
